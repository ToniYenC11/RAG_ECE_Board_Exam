# Knowledge Base

## Levels of Difficulty

| Symbol                | Meaning                                      | Type of Question                              |
| --------------------- | -------------------------------------------- | --------------------------------------------- |
| **K – Knowledge**     | Recall of facts, formulas, definitions       | Straight memorization questions               |
| **C – Comprehension** | Understanding concepts                       | Explain, interpret, classify                  |
| **A – Application**   | Applying formulas/techniques                 | Solve numerical or symbolic problems          |
| **A – Analysis**      | Breaking down problems, multi-step reasoning | Circuit analysis, decomposition, comparison   |
| **S – Synthesis**     | Combine concepts to form a solution          | System design, integration-type problems      |
| **E – Evaluation**    | Highest-order judgment                       | Assessing alternatives, verifying correctness |
